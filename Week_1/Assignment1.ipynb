```python
# 计算原始 horsepower 中位数
median_before = df["horsepower"].median()
print(median_before)
#Pandas 的 median() 默认会跳过 NaN 缺失值。
```

```python
# 找到 horsepower 的众数
mode_value = df["horsepower"].mode()[0]
print(mode_value)
```

```python
# 用众数填补 horsepower 的缺失值
df["horsepower"] = df["horsepower"].fillna(mode_value)
```

```python
# 计算填补后的中位数
median_after = df["horsepower"].median()
print("Before:", median_before)
print("After:", median_after)
```

```python
# Select two columns
selected = asia_car[["vehicle_weight", "model_year"]]
#因为你要同时选择多个列名,所以会有两层中括号
```

```python
#选择前7行，并创建一个7*2的矩阵
selected = selected.head(7)
selected
```

```python
X = selected.to_numpy()
print(X)
print(X.shape)
```

```python

# Matrix multiplication
XTX = X.T @ X
#其中@为矩阵乘法
print(XTX)
print(XTX.shape)
#因为X.T为2*7,(2*7)*(7*2)最后会生成2*2的矩阵
```

```python
# Inverse matrix
import numpy as np
XTX_inv = np.linalg.inv(XTX)
print(XTX_inv)
#np.linalg 是 NumPy 的线性代数模块。
#inv() 表示 inverse，即求逆矩阵。
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
# Calculate linear regression weights
w = XTX_inv @ X.T @ y
print(w)
print(w.sum())
```
